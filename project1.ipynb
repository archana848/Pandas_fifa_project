# Heighest salary player

```python
p[['Name','Salary']]
```

```python
p['Salary'] = p['Salary'].str.replace('€','')
p['Salary'] = p['Salary'].str.replace('K','000')
```

```python
p['Salary']
```

```python
p['Salary'] = p['Salary'].astype(int)
```

```python
p['Salary']
```

```python
high_salary = p['Salary'].max()
```

```python
high_salary
```

```python
highest_salary = p[p['Salary'] == high_salary][['Name','Salary']]
```

```python
highest_salary
```

```python
import matplotlib.pyplot as plt
```

```python
#plt.figure(figsize=(20,6))
#plt.plot(p['Name'], p['Salary'])
#plt.xlabel("Player Name")
#plt.ylabel("Salary (€)")
#plt.title("Player Salary Line Graph")
#plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.bar(highest_salary['Name'], highest_salary['Salary'])

plt.title("Highest Salary Player")
plt.xlabel("Player")
plt.ylabel("Salary")

plt.show()
```

# Lowest salary player

```python
p['Salary'].median()
```

```python
p['Salary'] = p['Salary'].replace(0, p['Salary'].median())
```

```python
low_salary = p['Salary'].min()
```

```python
low_salary
```

```python
lowest_salary = p[p['Salary'] == low_salary][['Name', 'Salary']]
```

```python
lowest_salary
```

```python
low5=lowest_salary.head(5)
```

```python
low5
```

```python
plt.figure(figsize=(6,4))
plt.bar(low5['Name'], low5['Salary'])
plt.title("Lowest Salary Player")
plt.xlabel("Player")
plt.ylabel("Salary")
plt.show()
```

# Tallest player

```python
p['Height']
```

```python
#tallest = p.sort_values(by="Height", ascending=False).head(1)
#tallest[["Name", "Height"]]
```

```python
p['Height'] = p['Height'].astype(str)
p["Height"] = p["Height"].str.replace("'", ".")
```

```python
p['Height'] = p['Height'].astype(float)
```

```python
tallest = p.loc[p["Height"] == p["Height"].max(), ["Name", "Height"]]
tallest
```

```python
plt.figure(figsize=(7,4))
plt.plot(tallest['Name'], tallest['Height'], marker='o')
plt.title("Tallest Players - Line Graph")
plt.xlabel("Player")
plt.ylabel("Height")
plt.grid(True)
plt.show()
```

# Shortest player

```python
shortest = p.loc[p["Height"] == p["Height"].min(), ["Name", "Height"]]
shortest
```

```python
sh5 = shortest.head(5)
```

```python
sh5
```

```python
plt.figure(figsize=(7,4))
plt.plot(sh5['Name'], sh5['Height'], marker='o')
plt.title("Shortest Players")
plt.xlabel("Player")
plt.ylabel("Height")
plt.grid(True)
plt.show()
```

# Top five highest salary players

```python
top5 = p.sort_values(['Salary'], ascending=False).head(5)[['Name','Salary']]
```

```python
top5
```

```python
plt.figure(figsize=(10,5))
plt.plot(top5['Name'],top5['Salary'],marker='o')
plt.title('Top 5 Highest Salary Players')
plt.xlabel('Players')
plt.ylabel('Salary')
plt.grid(True)
plt.show()
```
